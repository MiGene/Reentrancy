```python
import pandas as pd
from utils import find_nulls, recalculate_interactions, time_slice_df
```

```python
FOLD_1 = ['visor','akropolis','sanshu','dfx']
FOLD_2 = ['orion','cream','omni','noodle']
FOLD_3 = ['dao','revest','imbtc','fei']
FOLD_4 = ['spankchain','hypebear','ctoadz','bacon']
FOLD_5 = ['rari','lendfme','jay','weth']

FOLDS = [FOLD_1,FOLD_2,FOLD_3,FOLD_4,FOLD_5]
multiplier = 1.3
# FILL_SIZE = [500,500,500,500,500]
DATASET = '180'
if len(DATASET) > 0:
    DATASET = f'-{DATASET}'
FILES_PATH = f'../dataset/real-time{DATASET}/train_preprocessed'
FILLER_PATH = '../dataset/real-time/fold_filler'
SAVE_PATH = f'../dataset/real-time{DATASET}/fold'
PREFIX = 'realtime_'
# FOLD_SIZE = 400
```

```python
import os
os.makedirs(SAVE_PATH,exist_ok=True)
```

```python
def merge_files(folder_path,files_list,fold_num,prefix='',fill_size=None):
    sample_file_path = f'{folder_path}/{prefix}{files_list[0]}.csv'
    sample_df = pd.read_csv(sample_file_path)
    result_df = pd.DataFrame(columns=sample_df.columns)

    # Loop through each CSV file
    for file in files_list:
        file_path = f'{folder_path}/{prefix}{file}.csv'
        # Read the CSV file
        df = pd.read_csv(file_path)
        # Extract values from column A and append them to the result DataFrame
        result_df = pd.concat([result_df,df], ignore_index=True)
    return result_df
def fill_fold(fold_num,df,fill_size):
    filler_df = pd.read_csv(f'{FILLER_PATH}/{fold_num}.csv')

    # filler_df = filler_df.head(fill_size)
    filler_df = time_slice_df(filler_df,num_rows=fill_size)
    # filler_df = recalculate_interactions(filler_df)
    result_df = pd.concat([df,filler_df], ignore_index=True)

    return result_df
```

```python
dfs = []
df_size = []
for i in range(len(FOLDS)):
    fold_set = FOLDS[i]
    fold_df = merge_files(FILES_PATH,fold_set,i+1,PREFIX)
    dfs.append(fold_df)
    find_nulls(fold_df)
    df_size.append(fold_df.shape[0])
    
    print(len(fold_df))
max_size = max(df_size)
```

```python
import math

target_size = multiplier*max_size

for i in range(len(FOLDS)):
    df_to_fill = dfs[i]
    filled_fold = fill_fold(i+1,df_to_fill,math.floor(target_size-df_to_fill.shape[0]))
    save_path = f'{SAVE_PATH}/{i+1}.csv'
    filled_fold.to_csv(save_path,index=False)
```
