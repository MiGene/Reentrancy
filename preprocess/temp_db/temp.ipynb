```python
import glob
import os

import pandas as pd


DATASET_SUFFIX = ''
if len(DATASET_SUFFIX)>0:
    DATASET_SUFFIX = '-' + DATASET_SUFFIX
FOLDER_NAME = 'test_normal_preprocessed'
CSV_BASE_FOLDER = f'../../dataset/real-time{DATASET_SUFFIX}/{FOLDER_NAME}'


for file_path in glob.glob(os.path.join(CSV_BASE_FOLDER, "*.csv")):
    filename = os.path.basename(file_path)
    filename = filename.replace('test_','')
    df = pd.read_csv(file_path)
    print(filename)


    df.to_csv(filename,index=False)
```
