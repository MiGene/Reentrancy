```python
import os
import pandas as pd

folders = ['test_after_attack_now','test_attack','test_normal','train']

def count_nan(df,col):
    nan_count = df[col].isna().sum()
    return nan_count
    # print(f"Number of NaN values in {col}: {nan_count}")

def iter_files(folder):
    path = f'../dataset/real-time/{folder}'
    rows = 0
    csv_files = [file for file in os.listdir(path) if file.endswith('.csv')]
    total_depth = 0
    total_m_breadth = 0
    for file in csv_files:
        file_path = os.path.join(path, file)
        data = pd.read_csv(file_path)
        rows += len(data)
        total_depth += count_nan(data,'depth')
        total_m_breadth += count_nan(data,'max_breadth')
    return total_depth,total_m_breadth,rows

total_depth = 0
total_m_breadth = 0
rows = 0
for e in folders:
    d,b,r = iter_files(e)
    rows += r
    total_depth += d
    total_m_breadth += b
print(total_depth,total_m_breadth,rows)
```
